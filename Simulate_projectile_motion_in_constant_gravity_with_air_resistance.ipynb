```python
import math
import matplotlib.pyplot as plt
import numpy as np
from scipy.integrate import solve_ivp



totaltime= float(input("Total time?"))
vi= float(input("initlal velocity?"))
angle = math.radians(float(input("Angle in degrees?")))
drag = float(input("drag coefficient?"))
m = float(input("Mass of object?"))
g = 9.81


vx = vi*math.cos(angle)
vy = vi*math.sin(angle)


initial_states = [0,0,vx,vy]

def hit_ground(t, state):
    return state[1]

hit_ground.terminal = True
hit_ground.direction = -1

def position( t, state):
  x,y,vx,vy = state

  ax = -drag/m*vx
  ay = -drag/m*vy-g

  return[vx,vy,ax,ay]


solution = solve_ivp(
    position,
    [0, totaltime],
    initial_states,
    events=hit_ground,
    max_step=0.01)

x = solution.y[0]
y = solution.y[1]

plt.plot(x,y)
plt.xlabel("range")
plt.ylabel("height")
plt.title("range vs height with drag")
plt.show()
```
